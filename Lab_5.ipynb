Write a python program to perform **selection sort**.
- ** -Algorithm
* Step 1: Read the number of elements for the list from the user.
* Step 2: Using for loop insert the elements in the list.
* Step 3: Initialize the minimum element as min=numbers[i].
* Step 4: Using the swap method the elements are sorted accordingly.
* Step 5: Print the sorted list.

```python
#write code for above cell and compute the running time and also plot graph.
import time
import matplotlib.pyplot as plt
import random

def selection_sort(numbers):
    n = len(numbers)
    # Step 4: Sorting using the swap method
    for i in range(n):
        # Step 3: Initialize the minimum element
        min_idx = i
        for j in range(i + 1, n):
            if numbers[j] < numbers[min_idx]:
                min_idx = j

        # Swap the found minimum element with the first element
        numbers[i], numbers[min_idx] = numbers[min_idx], numbers[i]
    return numbers

def run_experiment():
    # Step 1: Read number of elements (We'll use a list of sizes for the graph)
    input_sizes = [100, 500, 1000, 1500, 2000, 2500, 3000]
    execution_times = []

    print("Starting Selection Sort Performance Analysis...")

    for size in input_sizes:
        # Step 2: Insert elements into the list (using random for testing)
        test_list = [random.randint(1, 10000) for _ in range(size)]

        # Compute running time
        start_time = time.time()
        selection_sort(test_list)
        end_time = time.time()

        duration = end_time - start_time
        execution_times.append(duration)
        print(f"Size: {size} | Time: {duration:.4f} seconds")

    return input_sizes, execution_times

# Execution and Plotting
sizes, times = run_experiment()

plt.figure(figsize=(10, 6))
plt.plot(sizes, times, marker='o', linestyle='-', color='orange', label='Selection Sort')
plt.title('Selection Sort: Execution Time vs Input Size')
plt.xlabel('Number of Elements (n)')
plt.ylabel('Time Taken (seconds)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Final Step 5: Print a sample sorted list
sample_list = [64, 25, 12, 22, 11]
print(f"\nExample Sorted List: {selection_sort(sample_list)}")
```

In computer science, **merge sort** (also commonly spelled mergesort) is an efficient, general-purpose, comparison-based sorting algorithm. Most implementations produce a stable sort, which means that the order of equal elements is the same in the input and output. Merge sort is a divide and conquer algorithm that was invented by John von Neumann in 1945.
Write a Python program to sort the elements using **merge sort** and plot a graph to the time taken versus n
- Algorithm(Merge Sort is a divide and conquer algorithm. It divides	input array in two halve.s, calls itself for the two halves and then merges two sorted halves.)
1.	Divide the unsorted array into n partitions, each	partition contains 1 element.
2.	Repeatedly merge partitioned units to produce	new sublists until there is only l sublist remaining. This will be the sorted	.
3.	Compare the first element of the sublist with the first element of the sublist to its right.
4.	Merge the two sublists by	paring each element of the sublist and placing the smaller element into the	new sublist.
5.	Repeat step 3 and 4 untill	all sublists are merged into a single sorted sublist.



```python
#write code for above cell and compute the running time and also plot graph.
import time
import matplotlib.pyplot as plt
import random

# 1. Implementation of Merge Sort
def merge_sort(arr):
    if len(arr) > 1:
        # Step 1: Divide the array into two halves
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]

        # Recursive calls for the two halves
        merge_sort(L)
        merge_sort(R)

        i = j = k = 0

        # Step 3 & 4: Compare elements and merge the two sublists
        while i < len(L) and j < len(R):
            if L[i] <= R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        # Step 5: Check if any element was left in L or R
        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1

# 2. Performance Measurement Setup
def measure_merge_sort():
    # We can use much larger input sizes because Merge Sort is efficient
    input_sizes = [1000, 5000, 10000, 20000, 40000, 60000, 80000, 100000]
    times = []

    print("Running Merge Sort Analysis...")
    for n in input_sizes:
        # Generate a random list of size n
        test_list = [random.randint(0, 100000) for _ in range(n)]

        start_time = time.perf_counter()
        merge_sort(test_list)
        end_time = time.perf_counter()

        duration = end_time - start_time
        times.append(duration)
        print(f"Size {n:6} | Time: {duration:.5f}s")

    return input_sizes, times

# 3. Plotting the results
sizes, execution_times = measure_merge_sort()

plt.figure(figsize=(10, 6))
plt.plot(sizes, execution_times, marker='o', linestyle='-', color='purple', label='Merge Sort')
plt.title('Merge Sort: Time Complexity Analysis ($O(n \log n)$)')
plt.xlabel('Input Size (n)')
plt.ylabel('Time Taken (seconds)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

**Quicksort** (sometimes called partition-exchange sort) is an efficient sorting algorithm. Developed by British computer scientist Tony Hoare in 1959 and published in 1961, it is still a commonly used algorithm for sorting.

Quicksort is a divide-and-conquer algorithm. It works by selecting a 'pivot' element from the array and partitioning the other elements into two sub-arrays, according to whether they are less than or greater than the pivot. The sub-arrays are then sorted recursively. This can be done in-place, requiring small additional amounts of memory to perform the sorting.
Write a Python program to sort the elements using quick sort and plot a graph to the time taken versus n.
- Algorithm
1.	Select a pivot element from the array.
2.	Partition the array into two sub-arrays. The elements in the first sub-array are less than the pivot element, while the elements in the second sub-array are greater than the pivot element.
3.	Recursively sort the sub-arrays created in Step 2.
4.	Join the sub-arrays and the pivot element together to	the sorted array.


```python
#write code for above cell and compute the running time and also plot graph.
import time
import matplotlib.pyplot as plt
import random
import sys

# Increase recursion depth for larger datasets
sys.setrecursionlimit(20000)

# 1. Implementation of Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        # Step 1: Select a pivot element (using the middle element)
        pivot = arr[len(arr) // 2]

        # Step 2: Partition the array
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]

        # Step 3 & 4: Recursively sort and join
        return quick_sort(left) + middle + quick_sort(right)

# 2. Performance Measurement Setup
def measure_quick_sort():
    # Testing sizes from 1,000 to 50,000
    input_sizes = [1000, 5000, 10000, 20000, 30000, 40000, 50000]
    times = []

    print("Running Quick Sort Analysis...")
    for n in input_sizes:
        # Generate random data for average-case performance
        test_list = [random.randint(0, 100000) for _ in range(n)]

        start_time = time.perf_counter()
        quick_sort(test_list)
        end_time = time.perf_counter()

        duration = end_time - start_time
        times.append(duration)
        print(f"Size {n:6} | Time: {duration:.5f}s")

    return input_sizes, times

# 3. Plotting the results
sizes, execution_times = measure_quick_sort()

plt.figure(figsize=(10, 6))
plt.plot(sizes, execution_times, marker='o', linestyle='-', color='teal', label='Quick Sort')
plt.title('Quick Sort: Execution Time vs Input Size ($O(n \log n)$)')
plt.xlabel('Input Size (n)')
plt.ylabel('Time Taken (seconds)')
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```

```python
# Experiment variables
n_list = [1000, 2000, 5000, 10000, 20000, 50000, 100000, 200000]
n_max = n_list[len(n_list) -1]
import time
import matplotlib.pyplot as plt
import random
import sys

# Increase recursion depth for Quick/Merge sort on large lists
sys.setrecursionlimit(200000)

# --- Sorting Algorithms ---

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr)//2
        L, R = arr[:mid], arr[mid:]
        merge_sort(L)
        merge_sort(R)
        i = j = k = 0
        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]; i += 1
            else:
                arr[k] = R[j]; j += 1
            k += 1
        while i < len(L):
            arr[k] = L[i]; i += 1; k += 1
        while j < len(R):
            arr[k] = R[j]; j += 1; k += 1

def quick_sort(arr):
    if len(arr) <= 1: return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

# --- Benchmarking Logic ---

n_list = [1000, 2000, 5000, 10000, 20000] # Reduced slightly as O(n^2) takes long at 200k
results = {"Selection": [], "Insertion": [], "Merge": [], "Quick": []}

for n in n_list:
    data = [random.randint(0, n) for _ in range(n)]

    for name, func in [("Selection", selection_sort), ("Insertion", insertion_sort),
                       ("Merge", merge_sort), ("Quick", quick_sort)]:
        test_data = data.copy()
        start = time.perf_counter()
        func(test_data)
        end = time.perf_counter()
        results[name].append(end - start)
        print(f"{name} Sort for n={n} took {end-start:.4f}s")

# --- Plotting ---

plt.figure(figsize=(12, 7))
for name, times in results.items():
    plt.plot(n_list, times, label=name, marker='o')

plt.title('Comparison of Sorting Algorithms')
plt.xlabel('Input Size (n)')
plt.ylabel('Time Taken (seconds)')
plt.legend()
plt.grid(True)
plt.show()
```
